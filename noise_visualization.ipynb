# Noise Analysis Visualization
Line plots with hue as Variable_Noised (Milk, RT, Water)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df = pd.read_csv('noise.csv')
print(df.head())
print(f"\nShape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
```

```python
# Set style
sns.set_style('whitegrid')
import numpy as np

colors = {'Milk': '#e74c3c', 'RT': '#3498db', 'Water': '#2ecc71'}
markers = {'Milk': 'o', 'RT': 's', 'Water': '^'}

# Plot 1: Cyber Anomalies (Bar plot - 3 bars per noise level)
fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(df['Noise_Level'].unique()))
width = 0.25
vars = sorted(df['Variable_Noised'].unique())

for i, var in enumerate(vars):
    data = df[df['Variable_Noised'] == var].sort_values('Noise_Level')
    ax.bar(x + i*width, data['Detected_Cyber_Anomalies'].values, width, 
           label=var, color=colors[var], alpha=0.8)

ax.set_xlabel('Noise Level', fontsize=12, fontweight='bold')
ax.set_ylabel('Detected Cyber Anomalies', fontsize=12, fontweight='bold')
ax.set_title('Cyber Anomalies Detection by Variable & Noise Level', fontsize=14, fontweight='bold')
ax.set_xticks(x + width)
ax.set_xticklabels(sorted(df['Noise_Level'].unique()))
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('plot_1_cyber_anomalies.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Saved: plot_1_cyber_anomalies.png")

# Plot 2: Biological Events (Line plot)
fig, ax = plt.subplots(figsize=(12, 6))
for var in df['Variable_Noised'].unique():
    data = df[df['Variable_Noised'] == var]
    ax.plot(data['Noise_Level'], data['Detected_Biological_Events'], 
            marker=markers[var], label=var, color=colors[var], linewidth=2.5, markersize=8)
ax.set_xlabel('Noise Level', fontsize=12, fontweight='bold')
ax.set_ylabel('Detected Biological Events', fontsize=12, fontweight='bold')
ax.set_title('Biological Events Detection by Variable', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('plot_2_biological_events.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Saved: plot_2_biological_events.png")

# Plot 3: Normal Detections (Line plot)
fig, ax = plt.subplots(figsize=(12, 6))
for var in df['Variable_Noised'].unique():
    data = df[df['Variable_Noised'] == var]
    ax.plot(data['Noise_Level'], data['Detected_Normal'], 
            marker=markers[var], label=var, color=colors[var], linewidth=2.5, markersize=8)
ax.set_xlabel('Noise Level', fontsize=12, fontweight='bold')
ax.set_ylabel('Detected Normal', fontsize=12, fontweight='bold')
ax.set_title('Normal Data Points Detection by Variable', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('plot_3_normal_detections.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Saved: plot_3_normal_detections.png")

# Plot 4: Combined comparison
fig, ax = plt.subplots(figsize=(12, 6))
for i, var in enumerate(vars):
    data = df[df['Variable_Noised'] == var].sort_values('Noise_Level')
    total_detected = data['Detected_Cyber_Anomalies'] + data['Detected_Biological_Events']
    ax.bar(x + i*width, total_detected.values, width, 
           label=var, color=colors[var], alpha=0.8)

ax.set_xlabel('Noise Level', fontsize=12, fontweight='bold')
ax.set_ylabel('Total Anomalies + Biological Events', fontsize=12, fontweight='bold')
ax.set_title('Total Event Detection by Variable & Noise Level', fontsize=14, fontweight='bold')
ax.set_xticks(x + width)
ax.set_xticklabels(sorted(df['Noise_Level'].unique()))
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('plot_4_total_events.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Saved: plot_4_total_events.png")

print("\n" + "="*50)
print("All individual plots saved successfully!")
print("="*50)
```
